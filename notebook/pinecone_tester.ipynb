```python
from pinecone import Pinecone
import os
key = os.getenv("PINECONE_API_KEY")
pc = Pinecone(api_key = key)

index = pc.Index('neural-note')
```

```python
from langchain_huggingface import HuggingFaceEmbeddings

embedding = HuggingFaceEmbeddings(model="all-MiniLM-l6-v2")
```

```python
from langchain_pinecone import PineconeVectorStore

vector_store = PineconeVectorStore(index=index, embedding=embedding)
```

```python
from langchain_classic.schema import Document


transcript = """Langgraph is a framework for agentic AI.
Langchain is a framework for gen ai.
Gen AI is generative ai provides text generation.
Agentic ai is an LLM backed system that can rationaly think and take actions.
"""
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

splitter = RecursiveCharacterTextSplitter(chunk_size=80, chunk_overlap=10)
chunks = splitter.split_text(transcript)
```

```python
docs = []

for i, chunk in enumerate(chunks):
    docs.append(
        Document(
            page_content=chunk,
            metadata={
                "chunk_id":i
            }
        )
    )
```

```python
docs
```

```python
vector_store.add_documents(docs)
```

```python
retriever = vector_store.as_retriever()
```

```python
query = "how does langgraph maintain state"

results = vector_store.similarity_search(query)
```
